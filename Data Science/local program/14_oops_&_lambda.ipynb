```python
class Employee:
    company = "TCS"

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def printDetail(self):
        print(f"the name is {self.name} of {self.company} with salary ₹{self.salary}")

    @staticmethod
    def printTime():
        print("the time is now")

    @classmethod
    def printClassDetail(cls):
        print(f"the company is {cls.company}")

e = Employee("Dharmik", 54200)
e.printDetail()
e.printTime()
e.printClassDetail()
```
