```python
import matplotlib.pyplot as plt

plt.style.use("ggplot")  # Choose any style you like
```

```python
labels = ['sachin', 'sehwag', 'king', 'yuvi']
runs = [18000, 15000, 21000, 11000]

plt.title("Pie Chart of Runs Scored by Players")
plt.pie(runs, labels=labels, autopct='%1.1f%%', startangle=90)
plt.axis('equal')  # Equal aspect ratio ensures that pie is drawn as a circle.
# plt.show()
```

```python
colors = ['#ff9999','#66b3ff','#99ff99','#ffcc99']
plt.pie(runs, labels=labels, colors=colors)
plt.show()
```

```python
plt.pie(
    runs,
    labels=labels,
    colors=colors,
    autopct='%1.1f%%',
    startangle=90,
    wedgeprops={'edgecolor': 'black', 'linewidth': 1.5, 'fill': True}
)
```

```python
explode = [0, 0, 0, 0.1]  # Only highlight Sachin's slice

plt.pie(runs, labels=labels, explode=explode, colors=colors, shadow=True,startangle=100, autopct='%1.2f%%',)
plt.title("Exploded Pie Example")
plt.show()
```
