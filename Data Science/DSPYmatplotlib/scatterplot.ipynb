```python
import matplotlib.pyplot as plt

study_hours = [1, 2, 3, 4, 5, 6, 7, 8, 9]
exam_scores = [40, 45, 90, 55, 60, 25, 75, 85, 90]

color = ['red' if score < 50 else 'green' for score in exam_scores]
size = [score * 2 for score in exam_scores]  # Size of points based on exam scores
social_scores = [12, 34, 67, 555, 678, 145, 56, 999, 445]

plt.scatter(study_hours, exam_scores, c=social_scores, cmap = 'viridis')
plt.colorbar(label='Exam Scores')  # Show color scale
plt.grid(True)

for i in range(len(study_hours)):
    plt.annotate(f's: {i+1}', (study_hours[i], exam_scores[i] - 2.5))

plt.title('Study Hours vs Exam Scores')
plt.xlabel('Study Hours')
plt.ylabel('Exam Scores')
plt.show()
```
