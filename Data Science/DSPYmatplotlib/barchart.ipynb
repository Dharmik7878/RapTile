```python
import matplotlib.pyplot as plt

years = [1990, 1992, 1994, 1996, 1998, 2000, 2003, 2005, 2007, 2010]
runs =  [500, 700, 1100, 1500, 1800, 1200, 1700, 1300, 900, 1500]

plt.bar(years, runs)
plt.xlabel("Year")
plt.ylabel("Runs Scored")
plt.title("Sachin Tendulkar's Yearly Runs")
```

```python
import numpy as np

sachin = [500, 700, 1100, 1500, 1800, 1200, 1700, 1300, 900, 1500]
sehwag = [0, 200, 900, 1400, 1600, 1800, 1500, 1100, 800, 0]
kohli  = [0, 0, 500, 800, 1100, 1300, 1500, 1800, 1900, 2100]

x = np.arange(len(years))  # the label locations
width = 0.25  # the width of the bars

plt.bar(x - width, sachin, width, label='Sachin')
plt.bar(x, sehwag, width = width, label='Sehwag') 
plt.bar(x + width, kohli, width, label='Kohli')

plt.xlabel("Year")
plt.ylabel("Runs Scored")
plt.title("Yearly Runs Comparison")
plt.xticks(x, years)
plt.legend()
plt.tight_layout()
```

```python
players = ["Sachin", "Sehwag", "Kohli", "Yuvraj"]
runs_5yrs = [500+700+1100+1500+1800, 0+200+900+1400+1600, 0+0+500+800+1100, 300+600+800+1100+900]

plt.barh(players, runs_5yrs)
plt.xlabel("Total Runs in First 5 Years")
plt.title("Total Runs Comparison in First 5 Years")
plt.tight_layout()
```

```python
import matplotlib.pyplot as plt

players = ["Sachin", "Sehwag", "Kohli"]
runs = [1500, 1200, 1800]

plt.bar(players, runs, color="#99eb95")

for i in range(len(players)):
    plt.text(i, runs[i] + 10, str(runs[i]), ha='center', va='bottom')
    
plt.xlabel("Players")
plt.ylabel("Runs Scored")
plt.title("Runs Scored by Players")
plt.tight_layout()
```
